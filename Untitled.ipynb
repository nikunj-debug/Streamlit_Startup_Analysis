```python
import pandas as pd
```

```python
startup=pd.read_csv('startup_cleaned.csv')
startup['city'] = startup['city'].replace({'Bengaluru': 'Bangalore', 'Gurgaon': 'Gurugram', 'New Delhi': 'Delhi'})
startup['date'] = pd.to_datetime(startup['date'],errors='coerce')
startup['month'] = startup['date'].dt.month
startup['year'] = startup['date'].dt.year
startup['startup']=startup['startup'].replace({'Flipkart.com':'Flipkart','Ola Cabs':'Ola','Olacabs':'Ola','Rapido Bike Taxi':'Rapido','Oyo Rooms':'OYO Rooms'})
startup['investors']=startup['investors'].replace({'SoftBank Group':'Softbank'})
```
