```python
import pandas as pd
```

```python
startup=pd.read_csv('startup_cleaned.csv')
startup['city'].replace({'Bengaluru':'Bangalore'},inplace=True)
startup['city'].replace({'New Delhi':'Delhi',},inplace=True)
startup['city'].replace({'Gurgaon':'Gurugram'},inplace=True)
startup['date'] = pd.to_datetime(startup['date'],errors='coerce')
startup['month'] = startup['date'].dt.month
startup['year'] = startup['date'].dt.year
```

```python
startup.info()
```

```python
df = startup[startup['investors'].str.contains('IDG')][['date', 'startup', 'vertical', 'city', 'round', 'amount']]
```

```python
# 1. Group by year and find the row index with the maximum amount for each year
max_per_year_idx = startup.groupby('year')['amount'].idxmax()

max_per_year = startup.loc[max_per_year_idx]
max_per_year[['year', 'startup', 'amount']].sort_values(by='year', ascending=False)
```
